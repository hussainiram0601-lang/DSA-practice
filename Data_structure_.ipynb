```python
n = 6
e = 7
edges = [(0,1),(0,3),(0,4),(1,2),(1,5),(2,4),(3,4)]
print(n,e,edges)
```

```python
adjecency_list = []
for i in range(n):
  adjecency_list.append([])
for edge in edges:
  x = edge[0]
  y = edge[1]
  adjecency_list[x].append(y)
  adjecency_list[y].append(x)
for i in range(n):
  print(i,"-->",adjecency_list[i])
```

```python
# adjacency matrix
ajdMatrix = []
for i in range(n):
  ajdMatrix.append([-1]*n)
for edge in edges:
  x = edge[0]
  y = edge[1]
  ajdMatrix[x][y] = 1
  ajdMatrix[y][x] = 1
for i in ajdMatrix:

   print(i)

```

```python
#BFS
# Using queue
class Queue:
  def __init__(self):
    self.q = []
    self.front = -1

  def push(self,x):
    if self.front == -1:
      self.front = 0
    self.q.append(x)

  def pop(self):
    if len(self.q) == 0:
      return -1
    x = self.q[self.front]
    self.front +=1
    if len(self.q) == self.front:
      self.front = -1
    return x

  def getFront(self):
    if len(self.q) == 0:
      return -1
    return self.q[self.front]

  def size(self):
    if self.front == -1:
      return 0
    return len(self.q) - self.front


q = Queue()
visited = [False]*n
ans = []

q.push(0)
visited[0] = True
ans.append(0)

while q.size()>0:
  front = q.pop()
  for i in adjecency_list[front]:
    if not visited[i]:
        q.push(i)
        visited[i] = True
        ans.append(i)
print(ans)

```

```python
#DFS using stack
class Stack:
  def __init__(self):
    self.st = []

  def push(self,x):
    self.st.append(x)

  def pop(self):
    if len(self.st) == 0:
      return -1
    x = self.st[-1]
    self.st.pop()
    return x

  def top(self):
    if len(self.q) == 0:
      return -1
    return self.st[-1]

  def size(self):
    return len(self.st)


st = Stack()
visited = [False]*n
ans = []

st.push(0)
visited[0] = True
ans.append(0)

while st.size()>0:
  top = st.pop()
  for i in adjecency_list[top]:
    if not visited[i]:
        st.push(i)
        visited[i] = True
        ans.append(i)
print(ans)
```

```python
# DFS using recurssion
visited = [False]*n
ans = []

def dfs(i , adjecency_list , visited):
  visited[i] = True
  ans.append(i)
  for x in adjecency_list[i]:
    if not visited[x]:
      dfs(x , adjecency_list , visited)
dfs(0 , adjecency_list , visited)
print(ans)
```

```python
# Detect cycle in an undirected graph
visited = [False]*n

def dfs(i ,parent, adjecency_list , visited):
  visited[i] = True
  for x in adjecency_list[i]:
    if x == parent:
      continue
    if visited[x]:
        return True
    if dfs(x ,i, adjecency_list , visited):
      return True
  return False
dfs(0 ,-1,  adjecency_list , visited)

```

```python
from heapq import heappush,heappop
n =  6
s = 0
edges = [(0,1,4),(0,2,4),(1,2,2),(2,3,3),(2,4,1),(2,5,6),(3,5,2),(4,5,3)]
# dijkatra's algorithm
adjecency_list = []
for i in range(n):
  adjecency_list.append([])
for edge in edges:
  x = edge[0]
  y = edge[1]
  w = edge[2]
  adjecency_list[x].append([y,w])
  adjecency_list[y].append([x,w])
visited = [False]*n
heap = []
dist = [float("inf")]*n
visited[s] = True
dist[s] = 0
heappush(heap,(dist[s],s))
for i in range(n):
  print(i,"-->",adjecency_list[i])
```
